## Getting stations

```python
import pandas as pd

filename_weather = "Synop-cleaned/clean_synop.csv"
filename_pollution = ""

weather_csv = pd.read_csv(filename_weather)
pollution_csv = pd.read_csv(filename_pollution)

weather_stations = weather_csv[
    ["geo_id_wmo", "name", "lat", "lon"]
].drop_duplicates()

pollution_station = pollution_csv[""].drop_duplicates()
```

```python
from sklearn.neighbors import BallTree
import numpy as np

weather_coords = np.radians(weather_stations["lat","long"])
pollution_coords = np.radians(pollution_station["",""])

tree = BallTree(weather_coords,metric="haversine")
```
